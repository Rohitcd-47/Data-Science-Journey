```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# 1. Define Column Groups based on Data Types & Business Logic

# Numerical columns that need scaling
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

# Ordinal column with a natural order/hierarchy
ordinal_cols = ['Contract']

# Categorical columns with no order (nominal / binary)
# Note: SeniorCitizen is already 0/1 numeric, so it passes through unchanged
nominal_cols = [
    'gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
    'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies', 'PaperlessBilling',
    'PaymentMethod'
]

# 2. Define Sub-Pipelines for Each Feature Type

# Numerical Pipeline: Impute missing values (if any) -> Standard Scale
num_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Ordinal Pipeline: Explicitly define category rank
# Order: 'Month-to-month' (0) < 'One year' (1) < 'Two year' (2)
ordinal_pipeline = Pipeline(steps=[
    ('ordinal_encoder', OrdinalEncoder(
        categories=[['Month-to-month', 'One year', 'Two year']]
    ))
])

# Nominal Pipeline: One-Hot Encode (drop='first' prevents collinearity/dummy variable trap)
nominal_pipeline = Pipeline(steps=[
    ('onehot_encoder', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
])

# 3. Combine into a Single Master ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('ord', ordinal_pipeline, ordinal_cols),
        ('nom', nominal_pipeline, nominal_cols)
    ],
    remainder='passthrough'  # Keeps 'SeniorCitizen' as 0/1 without modification
)

print("Preprocessor Pipeline created successfully!")
```

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

# ==========================================
# 1. CUSTOM EDGE-CASE SAFE TRANSFORMERS
# ==========================================
class RawDataCleaner(BaseEstimator, TransformerMixin):
    """
    Handles raw payload sanitization & edge cases:
    - Replaces empty strings/spaces in TotalCharges with 0.0
    - Coerces numerical columns to correct dtypes
    - Drops ID column safely if present
    """
    def __init__(self, id_col='customerID'):
        self.id_col = id_col

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        if self.id_col in X.columns:
            X = X.drop(columns=[self.id_col])
        
        # Clean TotalCharges edge case (space strings like ' ')
        if 'TotalCharges' in X.columns:
            X['TotalCharges'] = pd.to_numeric(X['TotalCharges'], errors='coerce').fillna(0.0)
        if 'tenure' in X.columns:
            X['tenure'] = pd.to_numeric(X['tenure'], errors='coerce').fillna(0)
        if 'MonthlyCharges' in X.columns:
            X['MonthlyCharges'] = pd.to_numeric(X['MonthlyCharges'], errors='coerce').fillna(0.0)
            
        return X

# ==========================================
# 2. FEATURE PIPELINE DEFINITION
# ==========================================
def build_production_pipeline():
    num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
    ordinal_cols = ['Contract']
    nominal_cols = [
        'gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
        'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies', 'PaperlessBilling',
        'PaymentMethod'
    ]

    num_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])

    ordinal_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('ordinal', OrdinalEncoder(
            categories=[['Month-to-month', 'One year', 'Two year']],
            handle_unknown='use_encoded_value', 
            unknown_value=-1
        ))
    ])

    nominal_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', num_pipeline, num_cols),
            ('ord', ordinal_pipeline, ordinal_cols),
            ('nom', nominal_pipeline, nominal_cols)
        ],
        remainder='passthrough'  # Keeps SeniorCitizen (0/1) intact
    )

    # Master End-to-End Pipeline
    full_pipeline = Pipeline([
        ('cleaner', RawDataCleaner()),
        ('preprocessor', preprocessor),
        ('classifier', RandomForestClassifier(
            n_estimators=200,
            max_depth=10,
            class_weight='balanced',
            random_state=42,
            n_jobs=-1
        ))
    ])
    
    return full_pipeline

# ==========================================
# 3. TRAINING & EXPORT EXECUTION
# ==========================================
if __name__ == "__main__":
    # Load Raw Dataset
    df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
    
    X = df.drop(columns=['Churn'])
    y = df['Churn'].map({'Yes': 1, 'No': 0})

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.20, random_state=42, stratify=y
    )

    # Train Complete Pipeline directly on RAW data
    pipeline = build_production_pipeline()
    pipeline.fit(X_train, y_train)

    # Evaluate
    y_pred = pipeline.predict(X_test)
    y_proba = pipeline.predict_proba(X_test)[:, 1]
    
    print("--- PRODUCTION PIPELINE EVALUATION ---")
    print(classification_report(y_test, y_pred))
    print(f"ROC-AUC Score: {roc_auc_score(y_test, y_proba):.4f}")

    # Export Single Executable Artifact
    joblib.dump(pipeline, "churn_model_pipeline.joblib")
    print("Model Artifact successfully saved as 'churn_model_pipeline.joblib'")
```

```python
import pandas as pd
import joblib

class ChurnPredictorService:
    def __init__(self, model_path="churn_model_pipeline.joblib"):
        # Load the single pipeline artifact containing cleaner, preprocessor, and model
        self.pipeline = joblib.load(model_path)
        self.required_columns = [
            'gender', 'SeniorCitizen', 'Partner', 'Dependents', 'tenure',
            'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
            'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
            'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod',
            'MonthlyCharges', 'TotalCharges'
        ]

    def predict(self, raw_input_dict: dict, threshold: float = 0.40) -> dict:
        """
        Takes raw API JSON input, validates fields, and returns prediction with risk assessment.
        """
        try:
            # Convert dictionary payload to DataFrame
            df_input = pd.DataFrame([raw_input_dict])

            # Edge Case Guard: Check missing required keys
            for col in self.required_columns:
                if col not in df_input.columns:
                    df_input[col] = np.nan

            # Predict Probability using loaded pipeline
            churn_probability = float(self.pipeline.predict_proba(df_input)[0, 1])
            
            # Custom Decision Threshold for High Recall in Business Context
            churn_prediction = int(churn_probability >= threshold)

            return {
                "status": "success",
                "churn_prediction": "Yes" if churn_prediction == 1 else "No",
                "churn_probability": round(churn_probability, 4),
                "risk_level": "High" if churn_probability >= 0.6 else ("Medium" if churn_probability >= 0.4 else "Low")
            }

        except Exception as e:
            return {
                "status": "error",
                "message": f"Inference pipeline failed: {str(e)}"
            }

# ==========================================
# TEST INFERENCE WITH RAW/DIRTY JSON PAYLOAD
# ==========================================
if __name__ == "__main__":
    predictor = ChurnPredictorService()

    # Raw dirty payload with edge cases (TotalCharges is blank space, missing customerID)
    sample_payload = {
        "gender": "Female",
        "SeniorCitizen": 0,
        "Partner": "Yes",
        "Dependents": "No",
        "tenure": 0,
        "PhoneService": "No",
        "MultipleLines": "No phone service",
        "InternetService": "DSL",
        "OnlineSecurity": "No",
        "OnlineBackup": "Yes",
        "DeviceProtection": "No",
        "TechSupport": "No",
        "StreamingTV": "No",
        "StreamingMovies": "No",
        "Contract": "Month-to-month",
        "PaperlessBilling": "Yes",
        "PaymentMethod": "Electronic check",
        "MonthlyCharges": 29.85,
        "TotalCharges": " "  # CRASH TEST: Blank space string
    }

    result = predictor.predict(sample_payload)
    print("API Prediction Output:")
    print(result)
```
