# **7. How is the correlation between cols?**

Correleation tell you how each column is related to or dependent to another column
Like if correlation is +ve then if one columns value increased related too increased
And if correlation is -ve then if one col increased another decrease

```python
df.corr(numeric_only=True)
```

```python
df.corr(numeric_only=True)['Survived']
```

See the correlation between age and survived is -ve
That means if age is less more survival chanes
