Q5. q5_bad_encoding.csv will throw a UnicodeDecodeError on a normal load. Figure out the actual encoding and load it without hardcoding the answer (detect it programmatically).

```python
import chardet, pandas as pd

with open("q5_bad_encoding.csv", "rb") as f:
    raw = f.read()
result = chardet.detect(raw)
print(result)   # {'encoding': 'ISO-8859-1', 'confidence': ...}

df = pd.read_csv("q5_bad_encoding.csv", encoding=result["encoding"])
print(df)
```
