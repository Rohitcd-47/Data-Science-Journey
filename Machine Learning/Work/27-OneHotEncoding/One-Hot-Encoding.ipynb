# **3. OneHotEncoding using Sklearn**

# **4. OneHotEncoding with Top Categories**

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon'),dtype=np.int32).sample(15)
```
