```python
import pandas as pd
import requests

df = pd.DataFrame()

for i in range(1, 11):
    response = requests.get("https://newsdata.io/api/1/latest?apikey=pub_4a383d9f98e0486a91ddfcd6d658b86a")
    
    temp_df = pd.DataFrame(response.json()['results'])[
        ['article_id', 'title', 'description', 'content', 'creator', 'language', 'country']
    ]
    
    df = pd.concat([df, temp_df], ignore_index=True)
```
